````python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

from wordcloud import WordCloud
import re, string, nltk
from sklearn.model_selection import train_test_split, GridSearchCV, cross_val_score
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.svm import LinearSVC
from sklearn.pipeline import Pipeline
from sklearn.metrics import (accuracy_score,classification_report,
confusion_matrix,ConfusionMatrixDisplay,
precision_score,recall_score,f1_score,roc_curve,auc)

nltk.download('stopwords')
nltk.download('wordnet')
from nltk.corpus import stopwords
from nltk.stem import WordNetLemmatizer
````

## Exploratory Data Analysis

````python
df['char_count']=df['message'].str.len()
df['word_count']=df['message'].str.split().str.len()

fig,ax=plt.subplots(1,2,figsize=(10,4))
sns.countplot(data=df,x='label',ax=ax[0])
sns.histplot(data=df,x='word_count',hue='label',bins=30,ax=ax[1])
plt.tight_layout()
plt.show()


````

## Preprocessing

````python
# Load the list of English stopwords (e.g., "the", "is", "and")
stop_words = set(stopwords.words('english'))

# Create a WordNet lemmatizer object to convert words to their base form
lemm = WordNetLemmatizer()

# Define a function to clean and preprocess SMS text
def clean_text(text):

    # Convert all characters to lowercase
    text = text.lower()

    # Remove URLs (e.g., http://example.com or www.example.com)
    text = re.sub(r'http\S+|www\S+', ' ', text)

    # Remove email addresses (e.g., abc@gmail.com)
    text = re.sub(r'\S+@\S+', ' ', text)

    # Remove all numeric digits from the text
    text = re.sub(r'\d+', ' ', text)

    # Remove punctuation marks such as !, ?, ., ,, :, etc.
    text = text.translate(str.maketrans('', '', string.punctuation))

    # Split the sentence into words,
    # remove stopwords,
    # and convert each word to its base (lemma) form
    tokens = [
        lemm.lemmatize(word)
        for word in text.split()
        if word not in stop_words
    ]

    # Join the cleaned words back into a single string
    return ' '.join(tokens)

# Remove duplicate SMS messages from the dataset
df = df.drop_duplicates()

# Apply the clean_text() function to every message
# and store the cleaned text in a new column named 'clean_message'
df['clean_message'] = df['message'].apply(clean_text)

# Convert categorical labels into numerical values
# ham  -> 0
# spam -> 1
df['label'] = df['label'].map({
    'ham': 0,
    'spam': 1
})

# Display the first five rows of the processed dataset
df.head()
````

## Train/Test Split

````python
from sklearn.model_selection import train_test_split

X = df['clean_message']
y = df['label']

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y
)

print(X_train.shape)
print(X_test.shape)
````

## SVM Model

````python
pipeline=Pipeline([
('tfidf',TfidfVectorizer(max_features=5000,ngram_range=(1,2))),
('svm',LinearSVC())
])

pipeline.fit(X_train,y_train)
pred=pipeline.predict(X_test)

print("Accuracy:",accuracy_score(y_test,pred))
print("Precision:",precision_score(y_test,pred))
print("Recall:",recall_score(y_test,pred))
print("F1:",f1_score(y_test,pred))
print(classification_report(y_test,pred))


````

### Confusion Matrix and Error Analysis

````python
# Plot Confusion Matrix
cm = confusion_matrix(y_test, pred)
disp = ConfusionMatrixDisplay(confusion_matrix=cm, display_labels=['Ham', 'Spam'])

fig, ax = plt.subplots(figsize=(6, 5))
disp.plot(cmap='Blues', ax=ax)
plt.title('Confusion Matrix for SVM Classifier')
plt.grid(False)
plt.show()
````

### Feature Importance: Top Predictors for Ham and Spam
Let's analyze which words the Linear Support Vector Machine associates most strongly with Spam versus Ham by inspecting the model coefficients.

````python
# Extract the feature names and coefficients
feature_names = pipeline.named_steps['tfidf'].get_feature_names_out()
coefficients = pipeline.named_steps['svm'].coef_[0]

coef_df = pd.DataFrame({
    'Feature': feature_names,
    'Coefficient': coefficients
})

# Sort to find top indicators
top_spam = coef_df.sort_values(by='Coefficient', ascending=False).head(15)
top_ham = coef_df.sort_values(by='Coefficient', ascending=True).head(15)

fig, axes = plt.subplots(1, 2, figsize=(15, 6))

# Plot Top Spam Predictors
sns.barplot(data=top_spam, x='Coefficient', y='Feature', ax=axes[0], palette='Reds_r')
axes[0].set_title('Top 15 Predictors for SPAM')
axes[0].set_xlabel('Coefficient Value')

# Plot Top Ham Predictors
sns.barplot(data=top_ham, x='Coefficient', y='Feature', ax=axes[1], palette='Blues')
axes[1].set_title('Top 15 Predictors for HAM')
axes[1].set_xlabel('Coefficient Value')

plt.tight_layout()
plt.show()
````

### Training Additional Models (Random Forest, LightGBM, KNN) and Comparison
Let's train Random Forest, LightGBM, and K-Nearest Neighbors (KNN) models using similar TF-IDF features, and compare their performance with the SVM model on our test dataset.

````python
import lightgbm as lgb
from sklearn.ensemble import RandomForestClassifier
from sklearn.neighbors import KNeighborsClassifier

# Define pipelines for the new models
rf_pipeline = Pipeline([
    ('tfidf', TfidfVectorizer(max_features=5000, ngram_range=(1,2))),
    ('rf', RandomForestClassifier(random_state=42, n_jobs=-1))
])

lgb_pipeline = Pipeline([
    ('tfidf', TfidfVectorizer(max_features=5000, ngram_range=(1,2))),
    ('lgb', lgb.LGBMClassifier(random_state=42, n_jobs=-1, verbose=-1))
])

knn_pipeline = Pipeline([
    ('tfidf', TfidfVectorizer(max_features=5000, ngram_range=(1,2))),
    ('knn', KNeighborsClassifier(n_neighbors=5, n_jobs=-1))
])

# Dictionary of models to evaluate (including our trained SVM pipeline)
models = {
    'Linear SVM': pipeline,
    'Random Forest': rf_pipeline,
    'LightGBM': lgb_pipeline,
    'KNN': knn_pipeline
}

# Train and predict for new models
for name, model_pipeline in models.items():
    if name != 'Linear SVM':
        model_pipeline.fit(X_train, y_train)
````

````python
results = []

for name, model_pipeline in models.items():
    test_preds = model_pipeline.predict(X_test)

    acc = accuracy_score(y_test, test_preds)
    prec = precision_score(y_test, test_preds)
    rec = recall_score(y_test, test_preds)
    f1 = f1_score(y_test, test_preds)

    results.append({
        'Model': name,
        'Accuracy': acc,
        'Precision': prec,
        'Recall': rec,
        'F1-Score': f1
    })

# Convert results to DataFrame
results_df = pd.DataFrame(results)
display(results_df.style.highlight_max(color='lightgreen', axis=0, subset=['Accuracy', 'Precision', 'Recall', 'F1-Score']))
````

````python
# Visualize the model comparison
results_melted = results_df.melt(id_vars='Model', var_name='Metric', value_name='Score')

plt.figure(figsize=(10, 6))
sns.barplot(data=results_melted, x='Metric', y='Score', hue='Model', palette='Set2')
plt.title('Performance Comparison of Classifiers')
plt.ylim(0.8, 1.02)
plt.ylabel('Score Value')
plt.grid(axis='y', linestyle='--', alpha=0.5)
plt.legend(bbox_to_anchor=(1.05, 1), loc='upper left')
plt.tight_layout()
plt.show()
````

### Detailed Error Analysis: False Positives and Negatives
Let's extract and view some of the misclassified messages to understand where the model makes errors.

````python
# Map indices back to get the original raw messages
test_analysis_df = pd.DataFrame({
    'Original_Message': df.loc[X_test.index, 'message'],
    'Clean_Message': X_test,
    'Actual_Label': y_test,
    'Predicted_Label': pred
})

false_positives = test_analysis_df[(test_analysis_df['Actual_Label'] == 0) & (test_analysis_df['Predicted_Label'] == 1)]
false_negatives = test_analysis_df[(test_analysis_df['Actual_Label'] == 1) & (test_analysis_df['Predicted_Label'] == 0)]

print(f"Total False Positives (Ham classified as Spam): {len(false_positives)}")
print(f"Total False Negatives (Spam classified as Ham): {len(false_negatives)}\n")

if not false_positives.empty:
    print("--- Sample False Positives ---")
    display(false_positives[['Original_Message', 'Clean_Message']].head(5))

if not false_negatives.empty:
    print("\n--- Sample False Negatives ---")
    display(false_negatives[['Original_Message', 'Clean_Message']].head(5))
````

## Hyperparameter Tuning

````python
param_grid={
'svm__C':[0.1,1,10]
}
grid=GridSearchCV(pipeline,param_grid,cv=5,n_jobs=-1)
grid.fit(X_train,y_train)

print("Best Parameters:",grid.best_params_)
print("Best CV Score:",grid.best_score_)
````

## Cross Validation

````python
scores=cross_val_score(grid.best_estimator_,df['clean_message'],df['label'],cv=5)
print(scores)
print("Mean Accuracy:",scores.mean())
````

## Prediction Function

````python
model=grid.best_estimator_

def predict_sms(text):
    pred=model.predict([clean_text(text)])[0]
    return "Spam" if pred==1 else "Ham"

print(predict_sms("Congratulations! You have won a free iPhone. Click here now."))
print(predict_sms("Hi, are we meeting tomorrow at 3 PM?"))
````

The goal of this project is to build a robust model capable of classifying SMS messages. The process involves:

## Key Steps

### 1. Data Loading and Initial Inspection
The `spam.csv` file is loaded, unnecessary columns are dropped, and columns are renamed for clarity. Basic checks for shape, info, null values, and duplicates are performed.

### 2. Exploratory Data Analysis (EDA)
-   Creation of `char_count` and `word_count` features.
-   Visualization of label distribution, word count distribution for ham vs. spam.

-   Definition of `clean_text` function to standardize messages.

The dataset is split into training and testing sets (80% train, 20% test) using `stratify=y` to maintain the proportion of ham/spam messages.

### 7. Cross Validation
Cross-validation is performed using the best estimator found during hyperparameter tuning to assess the model's generalized performance.

### 8. Prediction Function
A `predict_sms` function is provided to allow users to input a custom text message and receive a 'Spam' or 'Ham' classification.
